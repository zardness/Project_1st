```python
import pandas as pd
import glob
import os

# 1. 처리할 연도 설정 후 실행
target_year = '2024'

# 2. 경로 설정 (공통 폴더까지만 지정)
base_dir = "C:/data/"
folder_path = f"{base_dir}서울특별시 공공자전거 대여이력 정보_{target_year}/"

# 3. 해당 연도 폴더 안에서 전처리된 월별 파일들 찾기
all_files = glob.glob(os.path.join(folder_path, "cleaned_*_final.csv"))

if not all_files:
    print(f"⚠️ {target_year}년도 폴더에 파일이 없거나 경로를 찾을 수 없습니다.")
    print(f"확인한 경로: {folder_path}") # 경로가 잘 잡혔는지 확인용
else:
    print(f"📂 [{target_year}년] 병합 시작 (발견된 월별 파일: {len(all_files)}개)")
    
    # 4. 월별 파일들을 하나로 합치기
    year_dfs = []
    for f in all_files:
        df_month = pd.read_csv(f, encoding='cp949')
        year_dfs.append(df_month)
    
    # 데이터프레임 병합
    df_year_merged = pd.concat(year_dfs, ignore_index=True)
    
    # 5. 연도별 통합 파일로 저장
    output_path = f"{base_dir}total_{target_year}.csv"
    df_year_merged.to_csv(output_path, index=False, encoding='cp949')
    
    print(f"✅ [{target_year}년 완료] 저장 경로: {output_path}")
    print(f"   - 총 행 수: {len(df_year_merged):,}개, 열 수: {df_year_merged.shape[1]}개")
```
