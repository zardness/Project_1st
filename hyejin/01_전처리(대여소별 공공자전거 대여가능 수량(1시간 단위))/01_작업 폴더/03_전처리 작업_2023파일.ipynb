```python
import pandas as pd
import glob
```

```python
# 파일 경로 이름 줄이기 / 년 폴더 불러들이기
path = r'C:\ai\Source\09_Project\01_Data\서울특별시_대여소별 공공자전거 대여가능 수량(1시간 단위)\2023년'

# 컬럼 이름 변경
columns = ['일시', '대여소번호', '대여소명', '시간대', '거치대수량']

# 월 폴더 가져오기
file1 = f'{path}/data_231201~231215.csv'
file2 = f'{path}/data_231216~231231.csv'

df1 = pd.read_csv(file1, encoding='cp949', header=None, names=columns)
df2 = pd.read_csv(file2, encoding='cp949', header=None, names=columns)

# 두 데이터프레임 병합
df = pd.concat([df1, df2], ignore_index=True)
df
```

```python
# 전체 행렬 확인
df.shape
```

```python
# '대여소 번호' 컬럼 삭제
df.drop('대여소번호', axis=1, inplace=True)
```

```python
# 전체 행렬 확인
df.shape
```

```python
df.dtypes
```

```python
# 일시_datetime의로 유형 변경
df['일시'] = pd.to_datetime(df['일시'], format='%Y-%m-%d')
df.dtypes
```

```python
# 데이터 정렬 (일시 오름차순 -> 시간대 오름차순 순서로 정렬)
df = df.sort_values(by=['일시', '시간대'], ascending=[True, True])

# 정렬 후 뒤섞인 인덱스를 0부터 다시 깔끔하게 초기화
df = df.reset_index(drop=True)
```

```python
# 거치대 수량 결측치 확인
(df['거치대수량'] < 0).sum()
```

```python
# 요일 추가하기
df['요일'] = df['일시'].dt.dayofweek
df
```

```python
# 요일 위치 변경
df = df[['일시', '요일','시간대', '대여소명', '거치대수량']]
df
```

```python
# 파일 저장
df.to_csv('Cleaned/2023_Prep/2023.12_Prep-utf8.csv', encoding = 'utf-8-sig', index = False)
```
