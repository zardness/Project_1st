```python
import pandas as pd
```

```python
# 파일 경로 이름 줄이기 / 년도 폴더 가져오기
path = r'C:\ai\Source\09_Project\01_Data\서울특별시_대여소별 공공자전거 대여가능 수량(1시간 단위)\2025년'

# 월 파일 읽어들이기
df = pd.read_csv(f'{path}/data_2512.csv', encoding='cp949')
df
```

```python
# 전체 행렬 확인
df.shape
```

```python
# '대여소 번호' 컬럼 삭제
df.drop('대여소번호', axis=1, inplace=True)
```

```python
# 전체 행렬 확인
df.shape
```

```python
df.dtypes
```

```python
# 일시_datetime의로 유형 변경
df['일시'] = pd.to_datetime(df['일시'], format='%Y-%m-%d');
df.dtypes
```

```python
# 거치대 수량 결측치 확인
(df['거치대수량'] < 0).sum()
```

```python
# 요일 추가하기
df['요일'] = df['일시'].dt.dayofweek
df
```

```python
# 요일 위치 변경
df = df[['일시', '요일','시간대', '대여소명', '거치대수량']]
df
```

```python
# 파일 저장
df.to_csv('Cleaned/2025_Prep/2025.12_Prep-utf8.csv', encoding = 'utf-8-sig', index = False)
```
