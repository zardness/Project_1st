```python
def get_gu(addr):
    if pd.isna(addr):
        return None
    for token in str(addr).split():
        if token.endswith('구'):
            return token
    return None
```

```python
df_address['자치구'] = df_address['소재지(위치)'].apply(get_gu)
```

```python
gu_map = (
    df_address.drop_duplicates('보관소(대여소)명')
                .set_index('보관소(대여소)명')['자치구']
                .to_dict()
)
```

```python
df_data['대여 자치구'] = df_data['대여 대여소명'].apply(lambda x: gu_map.get(x))
df_data['반납 자치구'] = df_data['반납대여소명'].apply(lambda x: gu_map.get(x))
```

```python
df_data[df_data['대여 자치구'].isna()]['대여 대여소명'].unique()
```

```python
df_address['보관소(대여소)명'] = df_address['보관소(대여소)명'].str.strip()
df_data['대여 대여소명'] = df_data['대여 대여소명'].str.strip()
df_data['반납대여소명'] = df_data['반납대여소명'].str.strip()
```

```python
df_data['대여 자치구'] = df_data['대여 대여소명'].map(gu_map)
df_data['반납 자치구'] = df_data['반납대여소명'].map(gu_map)
```

```python
import re

def clean_name(x):
    return re.sub(r'^\d+\.\s*', '', str(x)).strip()

df_address['보관소(대여소)명'] = df_address['보관소(대여소)명'].apply(clean_name)
df_data['대여 대여소명'] = df_data['대여 대여소명'].apply(clean_name)
df_data['반납대여소명'] = df_data['반납대여소명'].apply(clean_name)
```
