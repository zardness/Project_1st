```python
df = pd.read_csv('C:/Users/Admin/Desktop/project/전처리완료/20/01_전처리완료.csv', encoding='cp949')
```

```python
# 대여 대여소명만 추출하고 중복 제거
rental_stations = df[['대여 대여소명']].drop_duplicates().copy()

print("2020년 1월 실제 이용된 대여소 수:", len(rental_stations))

rental_stations.head(20)
```

```python
return_stations = df[['반납대여소명']].drop_duplicates().copy()

print("2020년 1월 실제 이용된 반납 대여소 수:", len(return_stations))

return_stations.head(20)
```

```python
all_stations = pd.concat([
    df[['대여 대여소명']].rename(columns={'대여 대여소명': '대여소명'}),
    df[['반납대여소명']].rename(columns={'반납대여소명': '대여소명'})
]).drop_duplicates()

print("2020년 1월 전체 대여/반납 대여소 수:", len(all_stations))

all_stations.head(20)
```

```python
station_master = all_stations.copy()

station_master['자치구'] = None
```

```python
station = pd.read_csv('C:/Users/Admin/Desktop/project/data/공공자전거 대여소 정보(26.6월 기준).CSV', encoding='cp949')
station = station.drop(['대여소\n번호', 'Unnamed: 3','Unnamed: 4', 'Unnamed: 5','설치\n시기', '설치형태', 'Unnamed: 8','운영\n방식', 'Unnamed: 10'],axis=1)
station.info()
```

```python
print(station.head())
print(station.columns)
print(station.shape)
```

```python
station = station.rename(columns={
    '보관소(대여소)명': '대여 대여소명',
    '소재지(위치)': '자치구'
})
```
