```python
import pandas as pd
```

```python
df = pd.read_csv("./data_set/공공자전거_대여반납_이력정보/21/2112_merge.csv", encoding="utf-8-sig")
```

```python
# 1) 미확인 대여소 확인 (건수 많은 순)
r = df.loc[df["대여자치구"] == "미확인", "대여 대여소명"].value_counts().rename("대여 미확인")
b = df.loc[df["반납 자치구"] == "미확인", "반납 대여소명"].value_counts().rename("반납 미확인")
print(pd.concat([r, b], axis=1).fillna(0).astype(int))
```

```python
# 2) 대여소명 -> 자치구 (위 표를 보고 직접 확인해서 채우기)
fix = {
    "상암센터": "마포구",
    "중랑센터": "성동구",
    "이수센터": "서초구",
    "영남배송": "영등포구",
    "사대문센터": "종로구"  # 예시 (실제 위치 확인 후 수정하세요)
}
```

```python
# 3) 대여/반납 양쪽 모두 replace
for gu_col, name_col in (("대여자치구", "대여 대여소명"), ("반납 자치구", "반납 대여소명")):
    mask = (df[gu_col] == "미확인") & df[name_col].isin(fix)
    df.loc[mask, gu_col] = df.loc[mask, name_col].map(fix)
```

```python
print("남은 미확인 -> 대여:", (df["대여자치구"] == "미확인").sum(),
    "/ 반납:", (df["반납 자치구"] == "미확인").sum())
```

```python
# 4) 저장 (같은 이름으로 덮어씀. 원본을 남기려면 파일명을 바꾸세요)
df.to_csv("./data_set/공공자전거_대여반납_이력정보/21/2112_merge.csv", index=False, encoding="utf-8-sig")
```
