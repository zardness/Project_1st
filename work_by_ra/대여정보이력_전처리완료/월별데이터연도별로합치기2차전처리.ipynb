```python
import pandas as pd
import os

target_year = '2023'          # 👈 2021년이면 '2021', 2022년이면 '2022'로 변경
y_short = target_year[2:]     # 연도 뒤 2자리 자동 추출 (예: '20', '21')

# 폴더 경로 설정 (연도에 맞게 폴더명이 자동으로 바뀌도록 설정)
folder_path = f"C:/data/서울특별시 공공자전거 대여이력 정보_{target_year}/전처리완료{target_year}/"

print(f"=== [{target_year}년도] 월별 cleaned 파일 병합 시작 ===\n")
monthly_dfs = []

# ==========================================
# 📌 [2단계] 1월부터 12월까지 파일 불러오기
# ==========================================
for i in range(1, 13):
    m = f"{i:02d}"
    file_name = f"cleaned_{y_short}{m}.csv"  # 예: cleaned_2001.csv 형태
    file_path = os.path.join(folder_path, file_name)
    
    if os.path.exists(file_path):
        try:
            # 전처리된 파일 읽어오기 (인코딩 cp949)
            df_month = pd.read_csv(file_path, encoding='cp949', low_memory=False)
            
            # 날짜 컬럼 datetime 변환
            if '대여일시' in df_month.columns:
                df_month['대여일시'] = pd.to_datetime(df_month['대여일시'], errors='coerce')
            if '반납일시' in df_month.columns:
                df_month['반납일시'] = pd.to_datetime(df_month['반납일시'], errors='coerce')
                
            monthly_dfs.append(df_month)
            print(f"  - 로드 성공: {file_name} (행 수: {len(df_month):,})")
        except Exception as e:
            print(f"  - ❌ 로드 실패 ({file_name}): {e}")
    else:
        print(f"  - ⚠️ 파일 없음 (건너뜀): {file_name}")

# ==========================================
# 📌 [3단계] 전체 데이터 병합 및 최종 파일로 저장
# ==========================================
if monthly_dfs:
    yearly_df = pd.concat(monthly_dfs, ignore_index=True)
    
    # 연도별 이름으로 최종 저장 (예: merged_clean_2020.csv)
    output_filename = f"merged_clean_{target_year}.csv"
    output_path = os.path.join(folder_path, output_filename)
    yearly_df.to_csv(output_path, index=False, encoding='cp949')
    
    print(f"\n✅ [{target_year}년도] 최종 통합 완료!")
    print(f"  - 총 데이터 행 수: {len(yearly_df):,}")
    print(f"  - 저장 경로: {output_path}")
else:
    print(f"\n⚠️ [{target_year}년도] 병합할 cleaned 파일이 없습니다.")
```
