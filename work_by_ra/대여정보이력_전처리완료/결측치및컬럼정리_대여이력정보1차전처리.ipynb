```python
# 1. 작업용 데이터프레임 복사 (원본 df는 보존)
df_cleaned = df.copy()

# 2. 결측치(NaN 또는 \N 등)가 포함된 행 모두 제거
import numpy as np
df_cleaned.replace(r'\\N', np.nan, regex=True, inplace=True)
df_cleaned = df_cleaned.dropna()

# 3. 삭제할 컬럼 리스트 지정
cols_to_drop = [
    '자전거번호',
    '이용시간(분)', 
    '이용거리(M)', 
    '생년', 
    '성별', 
    '이용자종류', 
    '대여대여소ID',
    '반납대여소ID', 
    '자전거구분',
    '대여거치대', 
    '반납거치대'
]

# 4. 지정한 열들 삭제
df_cleaned.drop(columns=cols_to_drop, inplace=True, errors='ignore')

# 5. 결과 확인
print(f"최종 전처리된 데이터 크기 (행, 열): {df_cleaned.shape}")
print("남아있는 컬럼 목록:")
print(list(df_cleaned.columns))

# 6. 최종 전처리된 파일만 별도의 이름으로 저장
output_path = "C:/data/서울특별시 공공자전거 대여이력 정보_2024/cleaned_2401.csv"
df_cleaned.to_csv(output_path, index=False, encoding='cp949')

print(f"\n[완료] 전처리된 파일이 아래 경로에 저장되었습니다:\n{output_path}")
```

```python
import pandas as pd
import numpy as np

# 1. 처리할 연도 설정
target_year = '2024'
y_short = target_year[2:] # '20'

# 2. 처리할 월 리스트 정의 (1월부터 12월까지)
months = ['01', '02', '03', '04', '05', '06', '07', '08', '09', '10', '11', '12']

# 3. 삭제할 불필요한 컬럼 지정
cols_to_drop = [
    '자전거번호', '이용시간(분)', '이용거리(M)', '생년', '성별', 
    '이용자종류', '대여대여소ID', '반납대여소ID', '자전거구분', 
    '대여거치대', '반납거치대'
]

# 4. 해당 연도 폴더 경로 설정
folder_path = f"C:/data/서울특별시 공공자전거 대여이력 정보_{target_year}/"
print(f"=== [{target_year}년도] 월별 파일 전처리 및 대여소명 합치기 시작 (날짜 변환 제외) ===\n")

# 5. 1월부터 12월까지 반복문 실행
for m in months:
    file_name = f"서울특별시 공공자전거 대여이력 정보_{y_short}{m}.csv"
    file_path = folder_path + file_name
    
    print(f"[{target_year}년 {m}월 파일 처리 중] : {file_name}")
    
    try:
        # 데이터 로드 (인코딩 에러 방지 안전장치)
        try:
            df = pd.read_csv(file_path, encoding='cp949', low_memory=False)
        except UnicodeDecodeError:
            df = pd.read_csv(file_path, encoding='utf-8', low_memory=False)
            
        # 컬럼명 좌우 공백 제거
        df.columns = df.columns.str.strip()
        
        # 결측치(NaN 또는 \N) 제거
        df_final = df.copy()
        df_final.replace(r'\\N', np.nan, regex=True, inplace=True)
        df_final.dropna(inplace=True)
        
        # -------------------------------------------------------------
        # 대여소 번호 + 이름 합치기 (셀 제목: 대여대여소명)
        # -------------------------------------------------------------
        rent_no_col = '대여대여소번호' if '대여대여소번호' in df_final.columns else '대여 대여소번호'
        rent_name_col = '대여대여소명' if '대여대여소명' in df_final.columns else '대여 대여소명'
        
        df_final[rent_no_col] = df_final[rent_no_col].astype(str).str.strip()
        df_final[rent_name_col] = df_final[rent_name_col].astype(str).str.strip()
        df_final['대여대여소명'] = df_final[rent_no_col] + '.' + df_final[rent_name_col]
        
        if rent_name_col != '대여대여소명':
            df_final.drop(columns=[rent_name_col], inplace=True, errors='ignore')

        # -------------------------------------------------------------
        # 반납소 번호 + 이름 합치기 (셀 제목: 반납대여소명)
        # -------------------------------------------------------------
        return_no_col = '반납대여소번호' if '반납대여소번호' in df_final.columns else '반납 대여소번호'
        return_name_col = '반납대여소명' if '반납대여소명' in df_final.columns else '반납 대여소명'
        
        df_final[return_no_col] = df_final[return_no_col].astype(str).str.strip()
        df_final[return_name_col] = df_final[return_name_col].astype(str).str.strip()
        df_final['반납대여소명'] = df_final[return_no_col] + '.' + df_final[return_name_col]
        
        if return_name_col != '반납대여소명':
            df_final.drop(columns=[return_name_col], inplace=True, errors='ignore')

        # 불필요한 컬럼(거치대 포함) 및 원본 번호 컬럼 삭제
        total_cols_to_drop = cols_to_drop + [rent_no_col, return_no_col]
        df_final.drop(columns=total_cols_to_drop, inplace=True, errors='ignore')

        # 전처리 완료된 월별 파일 각각 저장 (cleaned_YYMM.csv)
        final_output_path = f"{folder_path}cleaned_{y_short}{m}.csv"
        df_final.to_csv(final_output_path, index=False, encoding='cp949')
        print(f"  - 완료 및 저장: {final_output_path} (행 수: {len(df_final):,})\n")

    except FileNotFoundError:
        print(f"  ❌ 에러: 파일을 찾을 수 없습니다. 경로를 확인해주세요: {file_path}\n")

print(f"=== [{target_year}년도] 날짜 변환 없이 모든 월별 파일 전처리 및 저장 완료! ===")
```
